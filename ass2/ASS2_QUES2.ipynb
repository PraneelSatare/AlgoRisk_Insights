```python
import yfinance as yf 

import numpy as np
import pandas as pd

def ans(stock_ticker,take_profit,stop_loss):
    data=yf.download(stock_ticker,start='2019-01-01',end='2024-01-01')
    data['Day']=range(1,len(data['Close'])+1)
    dates=np.array(data['Day'])
    closings=np.array(data['Close'])
    buy_list=[dates[0]]
    sell_list=[]
    profits=0
    losses=0
    for i in range(0,len(closings)):
        if(closings[i]>=(1+take_profit)*(closings[buy_list[-1]-1])):
            sell_list.append(dates[i])
            profits=profits+1
            if(i<len(closings)-1):
                buy_list.append(dates[i+1])
        elif(closings[i]<=(1-stop_loss)*(closings[buy_list[-1]-1])):
            sell_list.append(dates[i])
            losses=losses+1
            if(i<len(closings)-1):
                buy_list.append(dates[i+1])
    if(sell_list[-1]!=dates[-1]):
        sell_list.append(dates[-1])
    print('The buy list is :\n',buy_list,'\n')
    print('The sell list is:\n',sell_list) 
    print('Number of trades resluting in profit is:',profits)
    print('Number of trades resluting in loss is:',losses)
    
    
ans('INFY',0.04,0.02)
ans('ITC.NS',0.04,0.02)

```
