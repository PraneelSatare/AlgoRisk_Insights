```python
import matplotlib.pyplot as plt
import pandas as pd
def boolinger_bands(data,period,multiplier):
    data['cen']=data['Close'].rolling(window=period).mean()
    data['up']=data['cen']+multiplier*(data['Close'].rolling(window=period).std())
    data['low']=data['cen']-multiplier*(data['Close'].rolling(window=period).std())
    plt.plot(data['Close'])
    plt.plot(data['cen'],c='blue',lw=0.4)
    plt.plot(data['up'],c='red',lw=0.4)
    plt.plot(data['low'],c='green',lw=0.4)
    plt.show()
    
df=pd.read_csv('ndata.csv')
boolinger_bands(df,20,2)
```
