```python
import yfinance as yf 
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
def macd(data,sp,lp,sg):
#     sp=shorter period 
#     lp=longer period
#     sg=signal period
    data['macd']=data['Close'].ewm(span=sp,adjust="False").mean()-data['Close'].ewm(span=lp,adjust='False').mean()
    data['signal']=data['macd'].ewm(span=sg,adjust='False').mean()
    plt.plot(data['Date'],data['macd'])
    plt.plot(data['signal'])
    plt.show()
    
df=pd.read_csv('ndata.csv')
macd(df,12,26,9)
```
